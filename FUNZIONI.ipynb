### Come faccio a plottare il grafico di una funzione?

Per plottare (disegnare) il grafico di una funzione bisogna utilizzare il modulo Un modulo viene importato mediante la keyword <mark>matplotlib</mark>

Procedura

*   punto 1 : definisco la funzione
*   punto 2 : discretizzo la variabile x sul dominio
* punto 3 : scrivo la funzione determinando il valore di y per ogni x del dominio
* punto 4 : plotto la funzione utilizzando il modulo matplotlib



Esempio plotto la funzione
$$y = x^2$$

````python
# Punto 1 : definizione della funzione
def quadrato(x):
  """Ritorna il quadrato di `x`."""
  return x ** 2

# Punto 2 : discretizzazione del dominio
# devo importare il modulo numpy
import numpy
# utilizzo il metodo linspace(inzio, fine, passo)
x = numpy.linspace(0, 10, 100)
# punto 3 : scrivo la funzione
y = quadrato(x)
# punto 4: plotto la funzione
import matplotlib.pyplot
matplotlib.pyplot.plot(x, y)
matplotlib.pyplot.show()


````

### Lo stesso programma in C

```c
#include <stdio.h>

/* Punto 1: definizione della funzione */
double quadrato(double x)
{
    /* Ritorna il quadrato di x */
    return x * x;
}

int main(void)
{
    /* Punto 2: discretizzazione del dominio [0, 10] con 100 punti */
    const int n = 100;
    const double inizio = 0.0, fine = 10.0;
    double x[100], y[100];

    for (int i = 0; i < n; i++) {
        x[i] = inizio + i * (fine - inizio) / (n - 1);

        /* Punto 3: calcolo la funzione in ogni punto */
        y[i] = quadrato(x[i]);
    }

    /* Punto 4: il C non sa disegnare grafici da solo:
       stampo la tabella dei valori */
    printf("     x          y\n");
    for (int i = 0; i < n; i++) {
        printf("%8.3f   %8.3f\n", x[i], y[i]);
    }

    return 0;
}
```
